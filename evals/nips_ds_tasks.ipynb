```python
import json
import pandas as pd
import plotly.express as px


def main():
    try:
        # with open('aggregated_results.json', 'r') as f:
        with open(RESULT_FILE, "r") as f:
            data = json.load(f)
    except FileNotFoundError:
        print("Error: aggregated_results.json not found.")
        return

    # TARGET_TASKS = [
    #     "arc_challenge",
    # "arc_easy",
    # "hellaswag",
    # "lambada_openai",
    # "openbookqa",
    # "piqa",
    # "sciq",
    # "social_iqa",
    # "winogrande"
    # ]

    ratio_map = {"10p": "90%", "30p": "70%", "50p": "50%"}

    rows = []
    for task in TARGET_TASKS:
        if task not in data:
            continue

        task_data = data[task]
        for method, ratios in task_data.items():
            for ratio, tokens_dict in ratios.items():
                for token_count, metrics in tokens_dict.items():
                    accuracy = metrics.get("acc,none", metrics.get("acc_norm,none", 0))
                    num_tokens = int(token_count.replace("B", "").replace("T", ""))
                    display_ratio = ratio_map.get(ratio, ratio)

                    rows.append(
                        {
                            "Task": task,
                            "Method": method,
                            "Compression Ratio": display_ratio,
                            "Tokens": token_count,
                            "Num_Tokens": num_tokens,
                            "Accuracy": accuracy,
                        }
                    )

    df = pd.DataFrame(rows)
    if df.empty:
        return

    avg_df = df.groupby(
        ["Method", "Compression Ratio", "Tokens", "Num_Tokens"], as_index=False
    )["Accuracy"].mean()

    avg_df = avg_df.sort_values(by=["Compression Ratio", "Method", "Num_Tokens"])

    fig = px.line(
        avg_df,
        x="Tokens",
        y="Accuracy",
        color="Method",
        line_dash="Method",
        facet_col="Compression Ratio",
        markers=True,
        labels={"Accuracy": "Average Accuracy"},
        category_orders={
            "Tokens": ["1B", "2B", "4B", "6B"],
            "Compression Ratio": ["90%", "70%", "50%"],
            # Keep your standard, intended order for the legend
            "Method": ["PC", "PCD", "HP", "HPD"],
        },
        color_discrete_map={"PC": "red", "PCD": "red", "HP": "blue", "HPD": "blue"},
        line_dash_map={"PC": "dash", "PCD": "solid", "HP": "dash", "HPD": "solid"},
    )

    fig.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1]))

    # --- THE FIX ---
    # 1. Lock the current (correct) legend order by assigning a rank
    for i, trace in enumerate(fig.data):
        trace.legendrank = i

    # 2. Reverse the trace emission order so the first items (PC) are drawn last (on top visually)
    fig.data = fig.data[::-1]
    # ---------------

    fig.update_layout(
        height=600,
        width=1200,
        hovermode="x unified",
        template="plotly_white",
        legend_title_text="Method",
        font=dict(size=24),
        # No reversed traceorder hack needed here anymore
        legend=dict(font=dict(size=20)),
    )

    fig.update_traces(line=dict(width=3), marker=dict(size=10))

    fig.write_image("average_accuracy_chart.pdf", width=1200, height=600)
    fig.show()
    return fig


fig = main()
```

```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        # Calculate the middle index
        mid = (low + high) // 2
        guess = arr[mid]

        # Found the item
        if guess == target:
            return mid
        
        # The guess was too high
        if guess > target:
            high = mid - 1
        
        # The guess was too low
        else:
            low = mid + 1

    # Item does not exist in the list
    return -1


def bs(coll, e):
    l = 0
    h = len(coll) - 1

    while l <= h:
        m = int((l+h)/2)
        if coll[m] == e:
            return m
        if e < coll[m]:
            h = m - 1
        elif e > coll[m]:
            l = m + 1

import bisect

def bss(coll, e):
    # return bisect.bisect_right(coll, e)
    return bisect.


# Example Usage:
my_list = [1, 3, 5, 7, 9, 11, 13, 15]
print(f"Index of 7: {bss(my_list, 7)}")   # Output: 3
print(f"Index of 2: {bss(my_list, 2)}")   # Output: -1

my_list
```

```python
import hashlib


def get_bucket(input_string, num_buckets=10):
    # 1. Convert string to bytes and hash it
    hash_object = hashlib.sha256(input_string.encode())

    # 2. Convert the hex digest to an integer
    hash_int = int(hash_object.hexdigest(), 16)

    # 3. Use modulo to find the bucket (0-9)
    return hash_int % num_buckets


# Example usage:
item = "Asfsdgfdgdfffgdfgdfff"
bucket = get_bucket(item)
print(f"Item '{item}' goes into bucket: {bucket}")
```

```python
import binascii


def similarity_bucket(text, num_buckets=10):
    # This is a simplified "weighted" approach
    # Summing character codes creates a value relative to the content
    score = sum(ord(char) for char in text)

    # Using a simple linear scale makes 'Apples' and 'Apple' land close
    return (score // 10) % num_buckets


print(f"Apple: {similarity_bucket('Apple')}")  # Might be bucket 4
print(f"Apples: {similarity_bucket('Apples')}")  # Might be bucket 5
```

```python
def mergesort(left, right):
    i = 0
    j = 0
    sorted_arr = []
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            sorted_arr.append(left[i])
            i += 1
        else:
            sorted_arr.append(right[j])
            j += 1

    sorted_arr.extend(left[i:])
    sorted_arr.extend(right[j:])

    return sorted_arr


l = [4]
r = [1, 2, 4, 6]
t = mergesort(l, r).remove(4)
r.remove(4)
r
```
