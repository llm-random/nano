# MFU calculator
Give `solve` two of `mfu`, `time_h`, `steps`; it computes the third. Formulas live in `mfu_calc.py`.

```python
from mfu_calc import Setup

h = 16

setup = Setup(
    # model: h_balanced
    n_layers=h,
    dmodel=64 * h,
    dff=256 * h,
    q_heads=h,
    kv_heads=h // 2,
    ff_matrices=3,  # gated FF; 2 for GELU
    n_experts=1,
    topk=1,
    vocab=50_304,
    # run
    sl=256,
    bs=256,  # global batch, all GPUs together
    # hardware
    gpu="h100",  # h100 | gh200 | a100
    n_gpus=4,
    include_attention_scores=True,  # False reproduces the old 6 * N notebooks
)
setup.describe()
```

```python
# two of mfu (0-1), time_h (hours), steps -> the third
setup.solve(time_h=0.5, mfu=0.3)
```

```python
setup.solve(time_h=1, steps=20000)
```
