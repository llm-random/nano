```python
n_layers = 16
dmodel = 64 * n_layers
n_experts = 1
top_k = 1
dff = 2.5 * dmodel
kv_heads = n_layers
print(f"dmodel: {dmodel}")
print(f"2.5 dff: {2.5 * dmodel:.0f}\t4 dff {4 * dmodel:.0f}")
```

```python
attn_params = 2 * (1 + kv_heads / n_layers) * (dmodel ** 2)
ff_params = 3 * dmodel * dff * n_experts
ff_active = ff_params * top_k / n_experts
embedding = 50_304 * dmodel
model_params = n_layers * (attn_params + ff_params) + 2 * embedding
model_active = n_layers * (attn_params + ff_active) + embedding

print(f"active / total: {model_active / 1e9:.2f}B / {model_params / 1e9:.2f}B")
print(f"active / total: {model_active / 1e6:.2f}M / {model_params / 1e6:.2f}M")
```

```python
sl = 1024
bs = 64
# steps = 320_001
steps = 16_000
tokens = sl * bs * steps
print(f"tokens: {tokens / 1e9:.2f}B")
print(f"tok / param: {tokens / model_active:.2f}")
```
